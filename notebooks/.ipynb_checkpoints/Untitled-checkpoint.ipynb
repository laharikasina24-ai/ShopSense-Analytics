```python
import pandas as pd

df = pd.read_csv(
    r'..\data\raw\Global_Superstore.csv',
    encoding='latin1'
)

df.head()
```

```python
df.columns.tolist()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df['Order Date'] = pd.to_datetime(
    df['Order Date'],
    dayfirst=True
)

df['Ship Date'] = pd.to_datetime(
    df['Ship Date'],
    dayfirst=True
)
```

```python
df['Order Date'] = pd.to_datetime(
    df['Order Date'],
    format='%d-%m-%Y'
)

df['Ship Date'] = pd.to_datetime(
    df['Ship Date'],
    format='%d-%m-%Y'
)
```

```python
df[['Order Date', 'Ship Date']].head()
```

```python
df['Shipping Days'] = (
    df['Ship Date'] - df['Order Date']
).dt.days

df[['Order Date', 'Ship Date', 'Shipping Days']].head()
```

```python
df.isnull().sum()
df.duplicated().sum()
```

```python
df['Shipping Days'] = (
    df['Ship Date'] - df['Order Date']
).dt.days
```

```python
df.to_csv(
    r'..\data\cleaned\Global_Superstore_Cleaned.csv',
    index=False
)
```
